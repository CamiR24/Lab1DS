# Catch22 serie Marítima

```python
import pandas as pd
import pycatch22
import os
```

```python
def cargar_serie(path):
    s = pd.read_csv(path, index_col=0, parse_dates=True)['Viajero']
    s = s.asfreq('MS')
    return s

train_serie = cargar_serie('../../Datos/train_via_ingreso_maritima.csv')
print(f"Serie: Marítima | Train: {len(train_serie)} obs | {train_serie.index.min().date()} a {train_serie.index.max().date()}")

# --- Extraer las 22 caracteristicas de catch22 ---
resultado = pycatch22.catch22_all(train_serie.tolist(), catch24=False)
fila = dict(zip(resultado['names'], resultado['values']))
fila['serie'] = 'Marítima'

df_catch22 = pd.DataFrame([fila]).set_index('serie')
print(df_catch22.T)

os.makedirs('../../catch22_features', exist_ok=True)
df_catch22.to_csv('../../catch22_features/catch22_maritima.csv')
print("\nGuardado: ../../catch22_features/catch22_maritima.csv")
```
