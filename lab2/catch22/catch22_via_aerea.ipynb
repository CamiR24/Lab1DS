# Catch22 serie Vía Aérea

```python
import pandas as pd
import pycatch22
import os
```

```python
def cargar_serie(path):
    s = pd.read_csv(path, index_col=0, parse_dates=True)['Viajero']
    s = s.asfreq('MS')
    return s

full_serie = cargar_serie('../../Datos/via_ingreso_aerea.csv')
print(f"Serie: Vía Aérea | {len(full_serie)} obs | {full_serie.index.min().date()} a {full_serie.index.max().date()}")

# --- Extraer las 22 caracteristicas de catch22 ---
resultado = pycatch22.catch22_all(full_serie.tolist(), catch24=False)
fila = dict(zip(resultado['names'], resultado['values']))
fila['serie'] = 'Vía Aérea'

df_catch22 = pd.DataFrame([fila]).set_index('serie')
print(df_catch22.T)

# --- Guardar en carpeta compartida para armar la matriz despues ---
os.makedirs('../../catch22_features', exist_ok=True)
df_catch22.to_csv('../../catch22_features/catch22_via_aerea.csv')
print(f"\nGuardado: ../../catch22_features/catch22_via_aerea.csv")
```
