## 2.1 Explicación del catch 22

Catch22 (Canonical Time-series Characteristics) es un conjunto de 22 características diseñadas para resumir el comportamiento de una serie temporal con indicadores numéricos. Estas características describen diferentes propiedades de la serie, como su distribución, autocorrelación, complejidad, estacionalidad, variabilidad y patrones temporales, permitiendo representar cada serie con un vector de características en lugar de usar todos los valores originales.

Es importante porque facilita la comparación objetiva entre muchas series temporales, incluso cuando presentan diferentes escalas. Además, estas características pueden usarse como entrada para técnicas de aprendizaje automático y minería de datos, clustering y clasificación, ayudando a identificar similitudes, diferencias y patrones difíciles de encontrar solamente con el análisis exploratorio.

```python
import pandas as pd
import pycatch22
import os
```

```python
def cargar_serie(path):
    s = pd.read_csv(path, index_col=0, parse_dates=True)['Viajero']
    s = s.asfreq('MS')
    return s

serie = cargar_serie('../../Datos/san_cristobal.csv')
print(f"Serie: San Cristobal | {len(serie)} obs | {serie.index.min().date()} a {serie.index.max().date()}")

# --- Extraer las 22 caracteristicas de catch22 ---
resultado = pycatch22.catch22_all(serie.tolist(), catch24=False)
fila = dict(zip(resultado['names'], resultado['values']))
fila['serie'] = 'San Cristobal'

df_catch22 = pd.DataFrame([fila]).set_index('serie')
print(df_catch22.T)

# --- Guardar en carpeta compartida para armar la matriz despues ---
os.makedirs('../../catch22_features', exist_ok=True)
df_catch22.to_csv('../../catch22_features/catch22_san_cristobal.csv')
print("\nGuardado: ../../catch22_features/catch22_san_cristobal.csv")
```
